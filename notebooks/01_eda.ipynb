```python
WINDOWS = [30, 90]  # days
REGIMES = {
    # "pre-crisis": date(2019, 1, 1),
    # "gas crisis": date(2021, 7, 1),
    # "iberian cap": date(2022, 6, 15),
    # "post-cap": date(2024, 1, 1),
    "1h": omie_price_min_date,
    "15-min": date(2025, 10, 1),
}

regimes = pl.DataFrame({"regime": list(REGIMES), "date": list(REGIMES.values())})

# Add regimes
stats = omie_price.sort("datetime").join_asof(
    regimes.with_columns(pl.col("date").cast(pl.Datetime("us")).dt.replace_time_zone(MARKET_TZ)),
    left_on="datetime",
    right_on="date",
)

for w in WINDOWS:
    roll = dict(by="datetime", window_size=f"{w}d")
    stats = stats.with_columns(
        mean=pl.col(PRICE_ES).rolling_mean_by(**roll).over("regime"),
        std=pl.col(PRICE_ES).rolling_std_by(**roll).over("regime"),
    ).with_columns(
        resid=pl.col(PRICE_ES) - pl.col("mean"),
    ).with_columns(
        rmse=(pl.col("resid") ** 2).rolling_mean_by(**roll).sqrt().over("regime"), # Neg values
        resid_var=pl.col("resid").rolling_var_by(**roll).over("regime"),
    ).rename(lambda c: f"{c}_{w}d" if c in ("mean", "std", "resid", "rmse", "resid_var") else c)

# Added features
new_features = set(stats.columns) - set(omie_price.columns)
new_features.remove("date_right")

for part in stats.select(new_features).partition_by("regime"):
    show_df(
        part.describe(),
        caption=f"Stats for regime {part['regime'][0]}",
    )
```

- Prices appear to be cheaper (from 83 30D to 70 30D)
- Prices are more volatile. 30D std went from 28 on avg to 47. Almost double. Good for PnL -> more variance. 
- RMSE > std so prices are drifting. Less important for 15min. Little trend + noise. 
- Std / Mean: around 34% for hourly and 65% for 15min. Bigger spread available.

Stationarity analysis

```python
daily = stats.group_by(pl.col("datetime").dt.date().alias("date"), "regime").agg(pl.col(PRICE_ES).mean()).sort("date")


transforms = {
    "level": pl.col(PRICE_ES),
    "asinh": pl.col(PRICE_ES).arcsinh(), # Used for stabilizing variance (normal distribution), use due to negative values instead of logs
    "diff_1d": pl.col(PRICE_ES).diff(1),
    "diff_7d": pl.col(PRICE_ES).diff(7),
}

parts = {"all": daily} | {k[0]: v for k, v in daily.partition_by("regime", as_dict=True, maintain_order=True).items()}

show_df(
    pl.DataFrame(
        [
            {"regime": regime, "transform": name, **adf_kpss(part.select(expr).to_series())}

            # ADF: H0: non-stationary, H1: stationary
            # KPSS: H0: stationary, H1: non-stationary

            for regime, part in parts.items()
            for name, expr in transforms.items()
        ],
    ),
    precision=3,
)

# I need ADF p < 0.05 and KPSS p > 0.05 to conclude stationarity.

first_weeks = daily.filter(pl.col("date") < omie_price_min_date + timedelta(days=7 * 5))

fig = line_chart(
    first_weeks,
    "date",
    PRICE_ES,
    title="Daily mean prices (first five weeks)",
)
fig.add_scatter(
    x=first_weeks["date"],
    y=first_weeks[PRICE_ES].arcsinh(),
    mode="lines",
    name="asinh transform",
    yaxis="y2",
)
fig.add_scatter(
    x=first_weeks["date"],
    y=first_weeks[PRICE_ES].diff(1),
    mode="lines",
    name="1-day difference",
    yaxis="y2",
)
```

- Raw levels are obviously not stationary. 
- ADF is conclusive but kpss rejects stationarity. Looking at the chart, price seems rough but transformations seem to smooth. A little trend is probably causing the test to fail. 
- 1D differeencing is the most convincing with ADF and KPSS concluding stationarity.

Drift

```python
# PSI (Population Stability Index)
# Measures stability over time.
# PSI > 0.25 is a shift

rows = []
for regime, part in [
    ("1h", omie_price.filter(pl.col("datetime") < switch)),
    ("15-min", omie_price.filter(pl.col("datetime") >= switch)),
]:
    years = part["year"].unique().sort().to_list()
    for prev, y in zip(years, years[1:]):
        a = part.filter(pl.col("year") == prev)[PRICE_ES]
        b = part.filter(pl.col("year") == y)[PRICE_ES]
        rows.append({"regime": regime, "year": y, "psi": psi(a, b), "ks_stat": ks_2samp(a, b).statistic})

show_df(pl.DataFrame(rows))
```

- 5/7 years show shift. The data used to have a strong trend (probably Ukraine conflict). 

```python
hourly = omie_price.sort("datetime").group_by_dynamic("datetime", every="1h").agg(pl.col(PRICE_ES).mean()).with_columns(year=pl.col("datetime").dt.year())

# 1D differencing
hourly = hourly.with_columns(pl.col(PRICE_ES).diff(24).alias(PRICE_ES_DIFF_1D))

show_df(
    hourly.group_by("year")
    .agg(
        pl.col(PRICE_ES_DIFF_1D).skew().alias("skew"),
        pl.col(PRICE_ES_DIFF_1D).kurtosis().alias("kurt"),
        pl.col(PRICE_ES_DIFF_1D).quantile(0.01).alias("q01"),
        pl.col(PRICE_ES_DIFF_1D).median().alias("q50"),
        pl.col(PRICE_ES_DIFF_1D).quantile(0.99).alias("q99"),
        (pl.col(PRICE_ES_DIFF_1D) > 0).mean().alias("positive_share"),
    )
    .sort("year"),
)

histogram(omie_price, PRICE_ES, nbins=100, title="Distribution of day-ahead price ES")
```

```python
# Roughness ratio
# Roughness ratio = rolling std of 1D diff / rolling std of price

ROUGH_WINDOW = 24 * 7  # hours

rough = hourly.with_columns(
    (
        pl.col(PRICE_ES).diff().rolling_std(ROUGH_WINDOW)
        / pl.col(PRICE_ES).rolling_std(ROUGH_WINDOW)
    ).alias("roughness_ratio"),
)

r = pl.col("roughness_ratio")

show_df(
    rough.group_by("year")
    .agg(
        r.count().alias("count"),
        r.null_count().alias("null_count"),
        r.mean().alias("mean"),
        r.std().alias("std"),
        r.min().alias("min"),
        r.quantile(0.25).alias("25%"),
        r.median().alias("50%"),
        r.quantile(0.75).alias("75%"),
        r.max().alias("max"),
    )
    .sort("year")
    .with_columns(pl.col("year").cast(pl.String))
    .transpose(include_header=True, header_name="statistic", column_names="year"),
    caption="Roughness ratio by year",
)


carpet_plot(rough, "datetime", "roughness_ratio", size="small", title=f"Roughness ratio ({ROUGH_WINDOW}h window)")

rho = (1 - pl.col("roughness_ratio") ** 2 / 2).mean()
print(f"Average correlation with the previous hour: {rough.select(rho).item():.2f}")
```

Intuition:

- The average is 60 €/MWh and most hours sit between about 40 and 80, so std(price) ≈ 20.
- From one hour to the next, the price usually moves by around ±8, so std(Δprice) ≈ 8.
- Roughness = 8 / 20 = 0.4.
<br>

Interpretation:
- Price step is around 40% of the weekly price standard deviation.<br>
- Prices seem very persistent with a correlation of 0.9 with the previous hour.

The day-ahead auction for delivery day D closes at 12:00 CET on D-1. Participants submit sealed sell (V) and buy (C) bids for each of the 96 quarter-hours of D. OMIE, coupled with the other European markets, clears all bids at a single price per period, and every accepted bid is paid that price (pay-as-clear). Because of this, near-zero-cost producers (solar, wind, nuclear) bid at or below 0 €/MWh to make sure they are dispatched: about 17 GW here. Hydro and gas bid their opportunity and fuel costs, which form the rising part of the curve. On the demand side, retailers bid high prices to secure their customers' load, and storage and flexible consumers buy only at low prices. Offered supply (54 GW) far exceeds demand (28 GW) because most of that capacity is priced out. Only 23.3 GW cleared, at 169.02 €/MWh.

```python

month = loader.load("omie", "curva_pbc", last - timedelta(days=29), last).filter(
    pl.col("country").is_in(["MI", "ES"]),
    status="O",
).select("date", "period", "offer_type", "price_eur_mwh", "quantity_mw")

price_grid = pl.DataFrame({"price_eur_mwh": range(-500, 1501, 5)}, schema={"price_eur_mwh": pl.Float64})


def on_price_grid(side: str, descending: bool) -> pl.DataFrame:
    """Cumulative MW per period at each grid price."""
    steps = (
        month.filter(offer_type=side)
        .group_by("date", "period", "price_eur_mwh")
        .agg(pl.col("quantity_mw").sum())
        .sort("date", "period", "price_eur_mwh", descending=[False, False, descending])
        .with_columns(cum_mw=pl.col("quantity_mw").cum_sum().over("date", "period"))
        .sort("price_eur_mwh")
    )
    return (
        steps.select("date", "period").unique()
        .join(price_grid, how="cross")
        .sort("price_eur_mwh")
        .join_asof(
            steps.select("date", "period", "price_eur_mwh", "cum_mw"),
            on="price_eur_mwh",
            by=["date", "period"],
            strategy="forward" if descending else "backward",  # demand: buyers at >= p; supply: sellers at <= p
            check_sortedness=False,
        )
        .with_columns(pl.col("cum_mw").fill_null(0))
        .group_by("price_eur_mwh")
        .agg(
            mean=pl.col("cum_mw").mean(),
            lo=pl.col("cum_mw").quantile(0.025),
            hi=pl.col("cum_mw").quantile(0.975),
        )
        .sort("price_eur_mwh")
    )


fig = go.Figure()
for name, side, desc, color in [("supply (V)", "V", False, "#1f77b4"), ("demand (C)", "C", True, "#d62728")]:
    c = on_price_grid(side, desc)
    fig.add_scatter(
        x=c["lo"].to_list() + c["hi"].to_list()[::-1],
        y=c["price_eur_mwh"].to_list() + c["price_eur_mwh"].to_list()[::-1],
        fill="toself",
        fillcolor=color,
        opacity=0.2,
        line={"width": 0},
        name=f"{name} 95%",
        hoverinfo="skip",
    )
    fig.add_scatter(x=c["mean"], y=c["price_eur_mwh"], name=f"{name} mean", line={"color": color})
fig.update_layout(
    title=f"Bid curves, 30 days to {last}: mean and 95% band",
    xaxis_title="cumulative MW",
    yaxis_title="EUR/MWh",
)
fig.update_yaxes(range=[-50, 400])
fig
```

```python
# periods: "1".."25" (hourly) before Oct 2025, "H1Q1".."H25Q4" (15-min) after
hour = pl.col("period").str.extract(r"(\d+)").cast(pl.Int64)
quarter = pl.col("period").str.extract(r"Q(\d)").cast(pl.Int64).fill_null(1)
p, mw = pl.col("price_eur_mwh"), pl.col("quantity_mw")

# Spanish-side curve, as submitted: "MI" when coupled with PT, "ES" when the market splits
curve = (
    omie_curves.filter(
        pl.col("country").is_in(["MI", "ES"]),
        pl.col("offer_type").is_in(["V", "C"]),
        status="O",
    )
    .with_columns(
        datetime_utc=pl.col("date").cast(pl.Datetime("us")).dt.replace_time_zone(MARKET_TZ).dt.convert_time_zone("UTC")
        + pl.duration(hours=hour - 1, minutes=(quarter - 1) * 15),
    )
    .join(omie_price.select("datetime_utc", "datetime", "price_es"), on="datetime_utc")
)


def price_at_share(share: float) -> pl.Expr:
    """Price where cumulative MW hits share."""
    return p.sort().filter(mw.sort_by(p).cum_sum() / mw.sum() >= share).first()


sell = curve.filter(offer_type="V").group_by("datetime").agg(
    supply_mw=mw.sum(),
    supply_le0_mw=mw.filter(p <= 0).sum(),  # price takers: renewables, must-run
    supply_near_price_mw=mw.filter((p - pl.col("price_es")).abs() <= 10).sum(),  # flat = stable price
    complex_share=mw.filter(pl.col("offer_typology") != "S").sum() / mw.sum(),
    **{f"supply_p{int(s * 100)}": price_at_share(s) for s in (0.25, 0.5, 0.75, 0.9)},
)
buy = curve.filter(offer_type="C").group_by("datetime").agg(
    demand_mw=mw.sum(),
    demand_at_cap_mw=mw.filter(p >= p.max()).sum(),  # inelastic demand
)

curve_features = (
    sell.join(buy, on="datetime")
    .with_columns(margin_mw=pl.col("supply_mw") - pl.col("demand_mw"))
    .sort("datetime")
    .group_by_dynamic("datetime", every="1h")  # 15-min -> hourly mean, same grid as pre-Oct 2025
    .agg(pl.exclude("datetime").mean())
)

show_df(curve_features.describe(), caption="Bid-curve features (hourly)")

mw_cols = [c for c in curve_features.columns if c.endswith("_mw")]
price_cols = [c for c in curve_features.columns if c.startswith("supply_p")]
grid(
    [
        line_chart(curve_features, "datetime", mw_cols, title="Curve volumes (MW)"),
        line_chart(curve_features, "datetime", price_cols, title="Supply-curve price quantiles (EUR/MWh)"),
    ],
    rows=2,
    cols=1,
)
```

```python
CURVE_COLS = ["supply_mw", "demand_mw", "margin_mw", "supply_le0_mw", "supply_p50"]

curve_daily = (
    curve_features.group_by(pl.col("datetime").dt.date().alias("date"))
    .agg(pl.col(CURVE_COLS).mean())
    .sort("date")
    .join_asof(regimes, on="date")  # same "1h" / "15-min" regimes as prices
)

curve_transforms = {
    "level": lambda c: pl.col(c),
    "diff_1d": lambda c: pl.col(c).diff(1),
    "diff_7d": lambda c: pl.col(c).diff(7),  # removes the weekly cycle
}

curve_parts = {"all": curve_daily} | {
    k[0]: v for k, v in curve_daily.partition_by("regime", as_dict=True, maintain_order=True).items()
}

# ADF: H0 non-stationary. KPSS: H0 stationary.
# Stationary when ADF p < 0.05 and KPSS p > 0.05.
show_df(
    pl.DataFrame(
        [
            {"regime": regime, "feature": col, "transform": name, **adf_kpss(part.select(f(col)).to_series())}
            for regime, part in curve_parts.items()
            for col in CURVE_COLS
            for name, f in curve_transforms.items()
        ],
    )
    .with_columns(stationary=(pl.col("adf_p") < 0.05) & (pl.col("kpss_p") > 0.05))
    .sort("feature", "regime", "transform"),
    precision=3,
    caption="Stationarity of daily bid-curve features",
)

first_weeks = curve_daily.head(7 * 5)
fig = line_chart(first_weeks, "date", ["supply_mw", "demand_mw", "margin_mw"], title="Daily curve volumes (first five weeks)")
fig.add_scatter(
    x=first_weeks["date"],
    y=first_weeks["margin_mw"].diff(1),
    mode="lines",
    name="margin 1-day difference",
)
fig
```

Let's automate this analysis on all other features

```python
from datetime import date

import polars as pl

from src.stats.distribution_shift import psi
from src.stats.stationarity import adf_kpss

PSI_SHIFT = 0.25  # PSI above this = material year-over-year shift
LAGS = [24, 48, 168]  # hours, all >= 24 so usable day-ahead
T = "datetime_utc"


def to_hourly(df: pl.DataFrame) -> pl.DataFrame:
    """Common hourly grid, gaps as nulls."""
    return df.sort(T).group_by_dynamic(T, every="1h").agg(pl.col("value").mean()).upsample(T, every="1h")


def add_regimes(
    h: pl.DataFrame,
    regimes: dict[str, date],
) -> pl.DataFrame:
    if not regimes:
        return h.with_columns(regime=pl.lit("all"))
    starts = pl.DataFrame({"regime": list(regimes), "start": list(regimes.values())})
    return h.join_asof(
        starts.with_columns(pl.col("start").cast(h.schema[T])).sort("start"),
        left_on=T,
        right_on="start",
    ).drop("start")


def summarize(
    h: pl.DataFrame,
    target_diff: pl.DataFrame | None,
) -> dict[str, float | bool]:
    """Dynamics only; levels live in profile()."""
    x = pl.col("value")
    h = h.with_columns(
        diff_1d=x.diff(24),
        year=pl.col(T).dt.year(),
        mean_90d=x.rolling_mean_by(T, "90d"),
        std_90d=x.rolling_std_by(T, "90d"),
        rough=x.diff().rolling_std(168) / x.rolling_std(168),
    ).with_columns(
        rmse_90d=((x - pl.col("mean_90d")) ** 2).rolling_mean_by(T, "90d").sqrt(),
    )

    # stationarity on daily means
    daily = h.group_by_dynamic(T, every="1d").agg(x.mean())["value"]
    level, diff = adf_kpss(daily), adf_kpss(daily.diff())

    # year-over-year drift
    years = h["year"].unique().sort().to_list()
    by_year = {y: h.filter(pl.col("year") == y)["value"] for y in years}
    psis = [psi(by_year[a], by_year[b]) for a, b in zip(years, years[1:])]

    d = h["diff_1d"]
    row = {
        # ADF p < 0.05 and KPSS p > 0.05 -> stationary
        "level_stationary": level["adf_p"] < 0.05 and level["kpss_p"] > 0.05,
        "diff_1d_stationary": diff["adf_p"] < 0.05 and diff["kpss_p"] > 0.05,
        "years_shifted_share": sum(p > PSI_SHIFT for p in psis) / len(psis) if psis else None,
        "psi_max": max(psis, default=None),
        # > 1: trailing 90d mean lags the series -> drift, not just noise
        "rmse_over_std_90d": h["rmse_90d"].mean() / h["std_90d"].mean(),
        "roughness": h["rough"].median(),
        "rho_1h": 1 - h["rough"].median() ** 2 / 2,  # implied lag-1 correlation
        "acf_24h": h.select(pl.corr(x, x.shift(24))).item(),
        "acf_168h": h.select(pl.corr(x, x.shift(168))).item(),
        # tails of day-over-day changes
        "diff_1d_skew": d.skew(),
        "diff_1d_kurt": d.kurtosis(),  # excess, 0 = normal
        "diff_1d_q01": d.quantile(0.01),
        "diff_1d_q99": d.quantile(0.99),
    }
    if target_diff is not None:
        # does the feature's past change predict the target's change?
        joined = h.join(target_diff, on=T)
        for k in LAGS:
            row[f"xcorr_{k}h"] = joined.select(pl.corr(pl.col("diff_1d").shift(k), pl.col("target_diff_1d"))).item()
    return row


def feature_card(
    features: dict[str, pl.DataFrame],
    target: pl.DataFrame | None = None,
    regimes: dict[str, dict[str, date]] | None = None,
) -> pl.DataFrame:
    """One row per (feature, regime)."""
    target_diff = None if target is None else to_hourly(target).select(T, target_diff_1d=pl.col("value").diff(24))
    rows = []
    for name, df in features.items():
        h = add_regimes(to_hourly(df), (regimes or {}).get(name, {}))
        for part in h.partition_by("regime", maintain_order=True):
            rows.append({"series": name, "regime": part["regime"][0], **summarize(part, target_diff)})
    return pl.DataFrame(rows)
```

### ESIOS

```python
import polars as pl

def profile(frames: dict[str, pl.DataFrame]) -> pl.DataFrame:
    rows = []
    for name, df in frames.items():
        ts = df["datetime_utc"].sort()
        step = ts.diff().drop_nulls()
        rows.append({
            "series": name,
            "start": ts.min(),
            "end": ts.max(),
            "granularity": step.mode()[0],
            "pct_granularity_change": (step != step.mode()[0]).mean() * 100,
            "nan_pct": (df["value"].is_null() | df["value"].is_nan()).mean() * 100,
            **dict(df["value"].describe().iter_rows()),
        })
    return pl.DataFrame(rows)

def step_breakdown(frames: dict[str, pl.DataFrame]) -> pl.DataFrame:
    return pl.concat([
        df.sort("datetime_utc")
        .select(pl.lit(name).alias("series"), "datetime_utc", pl.col("datetime_utc").diff().alias("step"))
        .drop_nulls()
        .group_by("series", "step")
        .agg(
            pl.len().alias("n"),
            pl.col("datetime_utc").min().alias("first"),
            pl.col("datetime_utc").max().alias("last"),
        )
        .with_columns((pl.col("n") / pl.col("n").sum() * 100).alias("pct"))
        .sort("n", descending=True)
        for name, df in frames.items()
    ])
```

```python
esios = {v: loader.load("esios", v) for v in loader.variables("esios")}

for name, df in esios.items():
    print(f"{name:28} {df.shape}")

show_df(
    profile(esios)
)
show_df(
    step_breakdown(esios)
)
show_df(
    feature_card(esios)
)
```
